#IMPLEMENTATION OF BFS :

```python
from collections import deque

def bfs(graph, start):
    visited = set()
    queue = deque([start])

    visited.add(start)

    print("BFS Traversal:",end=" ")

    while queue:
        vertex = queue.popleft()
        print(vertex, end=" ")

        # Modified to use .get() for safe neighbor retrieval
        for neighbor in graph.get(vertex, []):
            if neighbor not in visited:
                visited.add(neighbor)
                queue.append(neighbor)

#Main Program
graph={}
n=int(input("Enter the number of vertices:"))
for i in range(n):
    vertex=input(f"Enter vertex{i+1}:")
    neighbors=input(f"Enter adjacent vertices of{vertex}(space-separated):").split()
    # Moved this line inside the loop to ensure all vertices are added
    graph[vertex]=neighbors


start=input("Enter the starting vertex:")
print("\nGraph:",graph)

bfs(graph,start)
```

#IMPLEMENTATION OF DFS :


```python
#Function for dfs
def dfs(graph,node,visited):
  if node not in visited:
    print(node,end=" ")
    visited.add(node)
  for neighbor in graph[node]:
    dfs(graph,neighbor,visited)

#Main Program
graph={}
n=int(input("Enter no. of vertices:"))
for i in range(n):
  vertex=input(f"Enter vertex{i+1}:")
  neighbor=input(f"Enter adjacent vertices of{vertex}(space-separated):").split()
  graph[vertex]=neighbor
start=input("Enter starting vertex:")
print("\nGraph:",graph)
```

#Hill-Climbing Algorithm


```python
def objective_function(x):
    return -x**2

def hill_climbing(start, step_size, max_iteration):
    current = start
    current_value = objective_function(current)

    for i in range(max_iteration):
        left = current - step_size
        right = current + step_size

        left_value = objective_function(left)
        right_value = objective_function(right)

        if left_value > current_value and left_value >= right_value:
            current = left
            current_value = left_value
        elif right_value > current_value:
            current = right
            current_value = right_value
        else:
            break

    return current, current_value


# Main program
start = float(input("Enter starting point: "))
step_size = float(input("Enter step size: "))
max_iteration = int(input("Enter max no. of iterations: "))

best_position, best_value = hill_climbing(
    start, step_size, max_iteration
)

print("Best position:", best_position)
print("Best value:", best_value)
```

#A* search Algorithm


```python
graph={
    'S':[('A',4),('B',3)],
    'A':[('C',5),('D',2)],
    'B':[('E',6),('F',4)],
    'C':[('G',3)],
    'D':[('G',2)],
    'E':[('G',1)],
    'F':[('G',2)],
    'G':[]
}
#HEURISTIC VALUES

heuristic={
    'S':8,
    'A':5,
    'B':4,
    'C':3,
    'D':2,
    'E':1,
    'F':2,
    'G':0
}
def astar(start,goal):
  open_list=[(start,0)]
  came_from={}
  g_cost={start:0}

  while open_list:
    #select node with min f=g+h
    current=min(open_list, key=lambda x:x[1]+heuristic[x[0]])
    open_list.remove(current)
    current_node=current[0]

    if current_node == goal:
      path=[goal]

      while current_node in came_from:
        current_node=came_from[current_node]
        path.append(current_node)

      path.reverse()
      return path,g_cost[goal]

    # This loop should be at the same indentation level as the 'if current_node == goal:' block
    for neighbor,cost in graph[current_node]:
      new_cost=g_cost[current_node]+cost
      if neighbor not in g_cost or new_cost<g_cost[neighbor]:
        came_from[neighbor]=current_node
        g_cost[neighbor]=new_cost
        open_list.append((neighbor, new_cost))

  # This return should be outside the 'while open_list' loop
  return None, float('inf')



#Main Program
start=input("Enter the start node:")
goal=input("Enter the goal node:")

path,cost=astar(start,goal)

if path:
  print("\nShortest Path:","->".join(path))
  print("Total Cost:",cost)
else:
  print("Path not found")
```

#Min-Max Algorithm

```python
def minimax(depth,nodeIndex,isMax,scores,height):
  #Base case:leaf node reached
  if (depth == height):
    return scores[nodeIndex]

  if isMax:
    return max(
        minimax(depth+1,nodeIndex*2,False,scores,height),
        minimax(depth+1,nodeIndex*2+1,False,scores,height)
    )
  else:
    return min(
        minimax(depth+1,nodeIndex*2,True,scores,height),
        minimax(depth+1,nodeIndex*2+1,True,scores,height)
    )


#Main Program
scores=list(map(int,input("Enter the 8 leaf node values(space-separated):").split()))
height=3

result=minimax(0,0,True,scores,height)
print("The optimal value is:",result)
```

#Alpha-Beta Pruning


**Aim**

To implement the Alpha-Beta Pruning Algorithm using python and obtain the optimal value from a game tree by eliminating unecessary braches.

**THEORY**

Alpha-Beta Pruning is an optimization technique used with the Min-Max Algorithm.It helps reduce the number of nodes that need to be evaluated while finding the optimal decision.

In Mon-Max Algorithm,the MAax player tries to maximize the score,while the MIN player tries to minimize the score.Min-Max may evaluate many branches even when some of them cannot affect the final decision.

Alpha-Beta Pruning avoids this unecessary computation by keeping track of 2 values:


*   **Alpha**:The best value that the MAX can gurantee so far
*   **Beta**:The best value that the MAX can gurantee so far

Initially:
 Alpha=-inf  
Beta=-inf

During search,if:beta<=alpha

the remaining nodes in tha6t branch are not explored,because they cannot influence the final decision.This is called **Pruning**.

Thus,Alpha-Beta Pruning produces the same optimal result as Min-Max, but it can do so by examining fewer nodes.

**Working Principle**
1. Start at Root node.
2. Explore the game tree using Min-Max aproach.
3. For Max nodes,select the maximum value and update Alpha.
4. For MIN nodes, select the minimum value and update Beta.
5. Whenever Beta<=Alpha,stop exploring the remaining nodes of that branch.
6. Continue until the optimal value is obtained.

**ALGORITM**
1. Start
2. Accept the Leaf-NOde values of the game tree as input.
3. Set:  
      Alpha=-inf  
      Beta=inf
4. Start the Alpha-Beta search from the Root node.
5. If a leaf node is reached ,return its value.
6. If the current node is Max node:  
Set best=-inf.  
Evaluate its child nodes.  
Update Alpha.  
If Beta<=Alpha, prune the remaining child nodes.
7. If the current node is MIN node:  
Set best=inf.  
Evaluate its child nodes.  
Update Beta.  
If Beta<=Alpha, prune the remaining child nodes.
8. Return the optimal value obtained at the Root node.
9. Display the optimal value.
10. Stop

```python
import math

#Alpha-Beta Pruning Function

def alpha_beta(depth, nodeIndex, maximizingPlayer, values, alpha, beta, height):
  #Base Case: Leaf Node Reached
  if depth==height:
    return values[nodeIndex]

  if maximizingPlayer:
    best=-math.inf

    for i in range(2):
      value=alpha_beta(
          depth+1,
          nodeIndex*2+i,
          False,
          values,
          alpha,
          beta,
          height
      )

      best=max(best, value)
      alpha=max(alpha, best)

      #Beta Cut-Off
      if beta<=alpha:
        break

    return best

  else:
    best=math.inf

    for i in range(2):
      value=alpha_beta(
          depth+1,
          nodeIndex*2+i,
          True,
          values,
          alpha,
          beta,
          height
      )

      best=min(best, value)
      beta=min(beta, best)

      #Alpha Cut-Off
      if beta<=alpha:
        break

    return best

#Main Program
values=list(map(int,input("Enter the 8 leaf node values(space-separated):").split()))
height=3

result=alpha_beta(0,0,True,values,-math.inf,math.inf,height)
print("The optimal value is:",result)
```

#8-Queens Problem

```python
def print_board(board):
  for row in board:
    print(" ".join(row))


#Function to check if a queen can be placed safely
def is_safe(board,row,col):
  #check column
  for i in range (row):
    if board[i][col]=='Q':
      return False
  #check upper-left diagonal
  i=row-1
  j=col-1

  while i>=0 and j>=0:
    if board[i][j]=='Q':
      return False
    i-=1
    j-=1

  #check upper-right diagonal
  i=row-1
  j=col+1

  while i>=0 and j<len(board):
    if board[i][j]=='Q':
      return False

    i-=1
    j+=1

  return True

#Backtracking Function
def solve(board,row):
  if row == 8:
    return True

  for col in range(8):
    if is_safe(board,row,col):
      board[row][col]='Q'

      if solve(board,row+1):
        return True
      else:
        board[row][col]='.'

  return False
#Main Program
board=[['.' for _ in range(8)]for _ in range(8)]

if solve(board,0):
  print("\nSolution for 8-Queens Problem:\n")
  print_board(board)
else:
  print("No solution exists.")
```

#Travelling Salesman Problem


```python
from itertools import permutations
#Functions to solve TSP
def travelling_salesman(graph,start):
  cities=list(range(len(graph)))
  cities.remove(start)

  min_cost=float('inf')
  best_path=None

  #Generate all possible tours
  for perm in permutations(cities):
    current_cost=0
    current_path=[start]
    k=start

    for city in perm:
      current_cost+=graph[k][city]
      current_path.append(city)
      k=city

    current_cost+=graph[k][start]
    current_path.append(start)

    if current_cost<min_cost:
      min_cost=current_cost
      best_path=current_path

  return best_path,min_cost


#Main Program
n=int(input("Enter no. of cities:"))

print("Enter cost of matrix:")
graph=[]

for i in range(n):
  row=list(map(int,input().split()))
  graph.append(row)

start=int(input("Enter the starting city(0 to {n-1}):"))

path,cost=travelling_salesman(graph,start)

print("\nOptimal Path:","->".join(map(str,path)))
print("Minimum Cost:",cost)
```
